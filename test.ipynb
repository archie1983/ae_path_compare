```python
from ae_path_compare import PathCompare
pc = PathCompare()
```

```python
path_refs = {}

path_refs["1"] = pc.load_images("/ae_path_compare/1/*.png")
path_refs["2"] = pc.load_images("/ae_path_compare/2/*.png")
path_refs["3"] = pc.load_images("/ae_path_compare/3/*.png")
path_refs["4"] = pc.load_images("/ae_path_compare/4/*.png")

#print(path_refs)
```

```python
def cmp_seq(cur_path, path_refs):
    pil_images = pc.load_images(cur_path)
    # compare that path against all reference paths
    cmp_res = {k: pc.fit_cur_path_to_ref_path(v, pil_images)[1] for k, v in path_refs.items()}
    print(cmp_res)

    best_match = max(cmp_res.items(), key=lambda k: k[1])
    response = {
        'best_match_ref': best_match[0],
        'best_match_score': best_match[1],
        'success': True
    }

    print(response)

cmp_seq("/ae_path_compare/tmp_cmp_A/*.png", path_refs)
cmp_seq("/ae_path_compare/tmp_cmp_B/*.png", path_refs)
cmp_seq("/ae_path_compare/tmp_cmp_C/*.png", path_refs)
cmp_seq("/ae_path_compare/tmp_cmp_D/*.png", path_refs)
```

```python
def fit_cur_path_to_ref_path(ref_path, cur_path):
    fittings = [pc.fit_single_img_to_ref_path(ref_path, cp) for cp in cur_path]
    ft = [f[1] if f[0] != 'U' else -1 for f in fittings]
    print(ft, fittings)

fit_cur_path_to_ref_path(path_refs['1'], pc.load_images("/ae_path_compare/tmp_cmp_B/*.png"))
```

```python
def fit_single_img_to_ref_path2(ref_path, img):
    probs = pc.compare_paths(ref_path, [img])
    print("Distribution 1 Analysis:")
    # print(confidence.analyze(dist1))
    # print("decision: ", confidence.agent_decision(confidence.analyze(dist1)))
    probs = probs.cpu().detach()
    print("probs: ", probs)
    #pc.visualize_probs(probs)
    dec = pc.confidence.agent_decision(pc.confidence.analyze(probs))
    return dec

def fit_cur_path_to_ref_path2(ref_path, cur_path):
    fittings = [fit_single_img_to_ref_path2(ref_path, cp) for cp in cur_path]
    ft = [f[1] if f[0] != 'U' else -1 for f in fittings]
    print(ft, fittings)

fit_cur_path_to_ref_path2(path_refs['1'], pc.load_images("/ae_path_compare/tmp_cmp_s2/*.png"))
```
